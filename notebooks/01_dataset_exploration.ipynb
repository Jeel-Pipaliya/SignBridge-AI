**Project:** SignBridge AI (Indian Sign Language Communication Platform)

### Objectives:
1. Scan the raw image dataset in `data/raw/`.
2. Inspect sample counts, missing/corrupted files, and class balance.
3. Check image dimensions and color channel consistency.
4. Visualize class distributions and representative sample image grids.

```python
import sys
from pathlib import Path

# Ensure project root is on Python path
PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import config
from data.ingest_dataset import scan_raw_dataset
from ai.dataset_exploration import compute_dataset_statistics, plot_class_distribution, plot_sample_grid

print("Project Root:", config.PROJECT_ROOT)
print("Raw Data Path:", config.RAW_DATA_DIR)
```

## 1. Scan Raw Dataset and Compute Class Statistics

```python
class_images = scan_raw_dataset(config.RAW_DATA_DIR)
stats = compute_dataset_statistics(class_images)

print(f"Total Classes:     {stats['total_classes']}")
print(f"Total Samples:     {stats['total_samples']}")
print(f"Imbalance Ratio:   {stats['imbalance_ratio']:.2f}")
print(f"Corrupted Files:   {stats['corrupted_count']}")
print(f"Image Resolutions: {stats['resolutions']}")

print("\nSample breakdown per class:")
for cls, count in sorted(stats['counts'].items()):
    print(f"  - Class '{cls}': {count} images")
```

## 2. Generate Class Distribution Bar Chart

```python
dist_plot_path = plot_class_distribution(stats['counts'])
print("Saved:", dist_plot_path)

from IPython.display import Image
Image(filename=str(dist_plot_path))
```

## 3. Generate Prototype Sample Grid

```python
grid_plot_path = plot_sample_grid(class_images)
print("Saved:", grid_plot_path)

Image(filename=str(grid_plot_path))
```
