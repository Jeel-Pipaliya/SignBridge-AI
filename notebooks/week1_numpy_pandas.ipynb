# NumPy and Pandas for SignBridge AI

Use this notebook to practice the array and dataframe operations that will later hold landmark sequences, numerical features, and CSV exports.

## Goals

- Understand NumPy arrays and shapes
- Practice Pandas dataframes and filtering
- Prepare for landmark-to-dataset conversion
- Keep the data model simple and inspectable.

```python
import numpy as np

arr = np.array([1, 2, 3, 4])
matrix = np.array([[1, 2], [3, 4]])
reshaped = arr.reshape(2, 2)

print('arr:', arr)
print('shape:', arr.shape)
print('dtype:', arr.dtype)
print('matrix:\n', matrix)
print('reshaped:\n', reshaped)
print('mean:', np.mean(arr))
print('std:', np.std(arr))
```

```python
import pandas as pd

frame = pd.DataFrame({
    'name': ['A', 'B', 'C'],
    'score': [80, 90, 70],
    'passed': [True, True, False],
})

print(frame.head())
print('\nInfo:')
print(frame.info())
print('\nDescribe:')
print(frame.describe())
print('\nFiltered rows:')
print(frame[frame['score'] > 80])
```

```python
from pathlib import Path

output_dir = Path('..') / 'data'
output_dir.mkdir(exist_ok=True)

csv_path = output_dir / 'sample_scores.csv'
frame.to_csv(csv_path, index=False)
loaded_frame = pd.read_csv(csv_path)

print(csv_path)
print(loaded_frame.to_dict(orient='records'))
```
